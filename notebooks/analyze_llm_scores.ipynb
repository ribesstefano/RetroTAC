## Reproducing the shared report's analysis

Recomputes, from `routes_llm_scores.csv` alone: the route / molecule-only /
in-stock split, whether a solved route predicts a better verdict (chi-square +
Kendall's tau), and per-category score/search-time distributions -- first for
the **routes** (Shallow-tree resolved a path), then separately for the
**molecules** (it didn't). Ten representative examples per category (spread
across the score range) and every flagged low-confidence/no-majority row are
shown for both.

```python
import json
import sys
import textwrap
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import display
from rdkit import Chem, RDLogger
from rdkit.Chem import Draw
from scipy import stats as sstats

RDLogger.DisableLog("rdApp.*")
sys.path.append(str(Path("..") / "src"))

import sys; sys.path.insert(0, "../scripts/llm_scoring")
from route_parsing import describe_molecule, parse_route_row

BRAND_COLOR = {"route": "#0EA068", "molecule_only": "#CC7F19", "in_stock": "#673489"}
ROUTE_CATS = [
    "Route feasible as it is",
    "Route feasible with few modifications",
    "Route feasible with significant modifications",
    "Route unfeasible",
    "Route was not solved to building blocks",
]
MOL_CATS = [
    "Readily synthesizable (standard building blocks & couplings)",
    "Synthesizable with routine medicinal-chemistry effort",
    "Challenging but feasible (multi-step / sensitive chemistry)",
    "Very challenging (long sequence / difficult selectivity or stereocontrol)",
    "Likely not synthesizable as drawn",
]
CAT_RANK = {c: i for i, c in enumerate(ROUTE_CATS)}
CAT_RANK.update({c: i for i, c in enumerate(MOL_CATS)})

AGREEMENT_FLOOR = 0.5   # <= this -> no clear ensemble majority ("disagreeing")
CONFIDENCE_FLOOR = 60   # <= this -> "low confidence" (roughly the bottom 5% overall)
N_REPS = 10

df["search_duration_s"] = pd.to_numeric(df["search_duration"], errors="coerce")  # "Exceeded" -> NaN
```

### How each row's score was generated

`parse_route_row` (in `scripts/llm_scoring/route_parsing.py`) reads the `resolved`
flag from a **Shallow-tree** retrosynthesis search and puts every row on one of
three paths before it ever reaches the LLM: a full retrosynthetic **route** to
grade, a **molecule-only** fallback when no route resolved, or an **in-stock**
short-circuit (target itself purchasable, 0 LLM calls, perfect score by
contract).

```python
mode_counts = df["llm_mode"].value_counts().reindex(["route", "molecule_only", "in_stock"])
mode_stats = pd.DataFrame({"n": mode_counts, "pct": (mode_counts / len(df) * 100).round(1)})
mode_stats
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(mode_stats.index, mode_stats["n"], color=[BRAND_COLOR[m] for m in mode_stats.index])
ax.set_ylabel("rows")
ax.set_title("How each target's score was generated")
for i, (n, pct) in enumerate(zip(mode_stats["n"], mode_stats["pct"])):
    ax.text(i, n, f"{n:,}\n({pct}%)", ha="center", va="bottom")
ax.margins(y=0.15)
plt.show()
```

### Run cost (from the raw per-sample JSONL)

`routes_llm_scores.csv` has no cost/token columns of its own, but its paired
`_raw.jsonl` (matched by row count against `logs/llm_scoring/llm_scoring_17056959.out`)
carries per-call `usage`/`cost` straight from litellm/OpenRouter -- ground truth
for what the run actually spent, not an estimate.

```python
raw_jsonl = Path("../data/llm_scoring/routes_with_score_20260711142743_raw.jsonl")

n_calls = cost = prompt_tokens = completion_tokens = 0
ensemble_sizes = {}
with open(raw_jsonl) as f:
    for line in f:
        rec = json.loads(line)
        k = rec.get("n_samples", 0)
        ensemble_sizes[k] = ensemble_sizes.get(k, 0) + 1
        for resp in rec.get("raw_responses", []):
            n_calls += 1
            usage = resp.get("usage") or {}
            cost += usage.get("cost") or 0.0
            prompt_tokens += usage.get("prompt_tokens") or 0
            completion_tokens += usage.get("completion_tokens") or 0

print(f"{n_calls:,} LLM calls -> ${cost:.2f}, {prompt_tokens + completion_tokens:,} tokens "
      f"({prompt_tokens:,} prompt + {completion_tokens:,} completion)")
print("ensemble size distribution (k actually achieved per row):", dict(sorted(ensemble_sizes.items())))
```

### Does a solved route predict a better verdict?

Map both category vocabularies onto one shared 0-4 feasibility rank (0=best) and
compare against `resolved` (did Shallow-tree find a path at all?): a chi-square
test of independence, and Kendall's tau for the ordinal correlation.

```python
rank = df["llm_category"].map(CAT_RANK)
solved = df["resolved"].astype(bool)
ok = rank.notna()

table = pd.crosstab(solved[ok], rank[ok])
chi2, chi_p, dof, _ = sstats.chi2_contingency(table)
tau, tau_p = sstats.kendalltau(solved[ok].astype(int), rank[ok])
print(f"chi2 = {chi2:.1f}   dof = {dof}   p = {chi_p:.2e}")
print(f"Kendall's tau = {tau:.3f}   p = {tau_p:.2e}   "
      "(negative: solved correlates with a BETTER, lower rank)")

pct = table.div(table.sum(axis=1), axis=0) * 100
present_ranks = sorted(int(r) for r in rank[ok].unique())
rank_label = {0: "best (0)", 1: "1", 2: "2", 3: "3", 4: "worst (4)"}

fig, ax = plt.subplots(figsize=(6, 3.4))
x = np.arange(len(present_ranks))
width = 0.36
ax.bar(x - width / 2, pct.loc[True].reindex(present_ranks, fill_value=0), width,
       label="solved", color=BRAND_COLOR["route"])
ax.bar(x + width / 2, pct.loc[False].reindex(present_ranks, fill_value=0), width,
       label="not solved", color=BRAND_COLOR["molecule_only"])
ax.set_xticks(x, [rank_label[r] for r in present_ranks])
ax.set_ylabel("% of group")
ax.set_xlabel("unified feasibility rank")
ax.set_title("Category rank distribution: solved vs. not solved")
ax.legend(frameon=False)
plt.show()
```

```python
def pick_representatives(cat_df: pd.DataFrame, mode: str, n: int = N_REPS) -> pd.DataFrame:
    """Spread-sample n rows across the score range, preferring ensemble-agreed /
    higher-confidence rows and (for routes) a legible step count. Falls back to
    the full category if too few rows clear those bars (e.g. every row flagged).
    """
    clean = cat_df[(cat_df["llm_category_agreement"] > AGREEMENT_FLOOR) &
                    (cat_df["llm_confidence"] > CONFIDENCE_FLOOR)]
    pool = clean if len(clean) > 0 else cat_df
    if mode == "route":
        legible = pool[pool["llm_n_steps"].between(1, 15)]
        if len(legible) >= min(n, len(pool)):
            pool = legible
    pool = pool.sort_values("llm_score").reset_index(drop=True)
    if len(pool) <= n:
        return pool
    idx = sorted(set(np.linspace(0, len(pool) - 1, n).round().astype(int)))
    return pool.iloc[idx]


def pick_flagged(cat_df: pd.DataFrame) -> pd.DataFrame:
    """Every row with no clear ensemble majority or low confidence -- not sampled."""
    mask = (cat_df["llm_category_agreement"] <= AGREEMENT_FLOOR) | (cat_df["llm_confidence"] <= CONFIDENCE_FLOOR)
    return cat_df[mask].sort_values("llm_score")


def show_representative(row: pd.Series, mode: str, idx: int) -> None:
    """Print the key fields (text-wrapped rationale) and render what the judge saw."""
    flags = []
    if row["llm_category_agreement"] <= AGREEMENT_FLOOR:
        flags.append("NO MAJORITY")
    if row["llm_confidence"] <= CONFIDENCE_FLOOR:
        flags.append("LOW CONFIDENCE")
    flag_str = f"  [{', '.join(flags)}]" if flags else ""
    print(f"--- #{idx}  score={row['llm_score']:.1f}  confidence={row['llm_confidence']:.0f}  "
          f"agreement={row['llm_category_agreement']:.0%}  solved={bool(row['resolved'])}{flag_str}")
    print(textwrap.fill(str(row["llm_rationale"]), width=100))

    display(Draw.MolToImage(Chem.MolFromSmiles(row["SMILES"]), size=(320, 260)))

    if mode == "route":
        parsed = parse_route_row(row.to_dict())
        rank_kind = {"T": 0, "I": 1, "B": 2, "U": 3}
        tags = sorted(parsed.tag_to_smiles, key=lambda t: (rank_kind[t[0]], int(t[1:]) if len(t) > 1 else 0))
        mols = [Chem.MolFromSmiles(parsed.tag_to_smiles[t]) for t in tags]
        display(Draw.MolsToGridImage(mols, molsPerRow=min(3, len(mols)), subImgSize=(320, 260), legends=tags))
        print(row["llm_route_description"])
    elif mode == "molecule_only":
        print(describe_molecule(row["SMILES"]))  # the RDKit descriptor line the judge saw
    print()


def show_flagged(flagged: pd.DataFrame) -> None:
    """Print every flagged row (no clear majority or low confidence) -- not a sample, all of them."""
    if flagged.empty:
        return
    print(f"\n{len(flagged)} flagged rows (no clear ensemble majority, or confidence <= {CONFIDENCE_FLOOR}), all shown:")
    for _, row in flagged.iterrows():
        flags = []
        if row["llm_category_agreement"] <= AGREEMENT_FLOOR:
            flags.append("no majority")
        if row["llm_confidence"] <= CONFIDENCE_FLOOR:
            flags.append("low confidence")
        snippet = textwrap.shorten(str(row["llm_rationale"]), width=140)
        print(f"  score={row['llm_score']:5.1f}  confidence={row['llm_confidence']:5.1f}  "
              f"agreement={row['llm_category_agreement']:.0%}  [{', '.join(flags)}]  {snippet}")


def analyze_category(cat_df: pd.DataFrame, mode: str, category: str) -> None:
    """Full drill-down for one (mode, category): 10 representatives + every flagged row."""
    reps = pick_representatives(cat_df, mode)
    flagged = pick_flagged(cat_df)
    print(f"\n{'=' * 90}\n[{mode}] {category}  "
          f"(n={len(cat_df):,}, mean score={cat_df['llm_score'].mean():.1f}, "
          f"mean confidence={cat_df['llm_confidence'].mean():.1f})\n{'=' * 90}")
    for i, (_, row) in enumerate(reps.iterrows(), 1):
        show_representative(row, mode, i)
    show_flagged(flagged)
```

## Section 1: Analyzing the routes

Every target below is **solved**: Shallow-tree found a path to purchasable
material -- whether via a multi-step disconnection or because the target was
already sitting in stock.

```python
route_pool = df[df["llm_mode"].isin(["route", "in_stock"])]

fig, ax = plt.subplots(figsize=(6, 3.6))
sns.histplot(x=route_pool["llm_score"], bins=20, kde=True, color=BRAND_COLOR["route"], ax=ax)
ax.set_xlabel("llm_score")
ax.set_title(f"Route-analysis score distribution (n={len(route_pool):,})")
plt.show()
```

```python
ok = route_pool["search_duration_s"].notna() & (route_pool["search_duration_s"] > 0)
d, s = route_pool.loc[ok, "search_duration_s"], route_pool.loc[ok, "llm_score"]
pear_r, pear_p = sstats.pearsonr(np.log10(d), s)
spear_r, spear_p = sstats.spearmanr(d, s)
print(f"Pearson r (log-time vs score) = {pear_r:.2f}  p={pear_p:.1e}")
print(f"Spearman rho (time vs score)  = {spear_r:.2f}  p={spear_p:.1e}   n={ok.sum():,}")

fig, ax = plt.subplots(figsize=(6, 3.6))
ax.scatter(d, s, s=10, alpha=0.35, color=BRAND_COLOR["route"])
ax.set_xscale("log")
ax.set_xlabel("search time, seconds (log scale)")
ax.set_ylabel("llm_score")
ax.set_title("Route-analysis: search time vs. score")
plt.show()
```

```python
route_cat_stats = (
    route_pool.groupby(["llm_mode", "llm_category"], dropna=False)
              .agg(n=("SMILES", "size"), mean_score=("llm_score", "mean"),
                   mean_confidence=("llm_confidence", "mean"), mean_agreement=("llm_category_agreement", "mean"))
              .reset_index()
)
route_cat_stats["pct_of_mode"] = route_cat_stats["n"] / route_cat_stats.groupby("llm_mode")["n"].transform("sum") * 100
route_cat_stats = (route_cat_stats.assign(_o=route_cat_stats["llm_category"].map(CAT_RANK))
                                   .sort_values(["llm_mode", "_o"]).drop(columns="_o").round(1).reset_index(drop=True))
route_cat_stats
```

```python
for mode in ["route", "in_stock"]:
    sub = route_pool[route_pool["llm_mode"] == mode]
    cats = ROUTE_CATS if mode == "route" else MOL_CATS[:1]
    for cat in cats:
        cat_df = sub[sub["llm_category"] == cat]
        if len(cat_df) == 0:
            continue
        analyze_category(cat_df, mode, cat)
```

## Section 2: Analyzing the molecules

Every target below is **not solved**: Shallow-tree's search never resolved a
path to purchasable building blocks, so the judge sizes up the bare structure
instead (RDKit descriptors + SMILES, no route).

```python
mol_pool = df[df["llm_mode"] == "molecule_only"]

fig, ax = plt.subplots(figsize=(6, 3.6))
sns.histplot(x=mol_pool["llm_score"], bins=20, kde=True, color=BRAND_COLOR["molecule_only"], ax=ax)
ax.set_xlabel("llm_score")
ax.set_title(f"Molecule-analysis score distribution (n={len(mol_pool):,})")
plt.show()
```

```python
ok = mol_pool["search_duration_s"].notna() & (mol_pool["search_duration_s"] > 0)
n_excluded = (~ok).sum()
d, s = mol_pool.loc[ok, "search_duration_s"], mol_pool.loc[ok, "llm_score"]
pear_r, pear_p = sstats.pearsonr(np.log10(d), s)
spear_r, spear_p = sstats.spearmanr(d, s)
print(f"Pearson r (log-time vs score) = {pear_r:.2f}  p={pear_p:.2f}")
print(f"Spearman rho (time vs score)  = {spear_r:.2f}  p={spear_p:.2f}   n={ok.sum():,}, "
      f"{n_excluded:,} excluded (timed out -- search_duration == 'Exceeded')")

fig, ax = plt.subplots(figsize=(6, 3.6))
ax.scatter(d, s, s=14, alpha=0.5, color=BRAND_COLOR["molecule_only"])
ax.set_xscale("log")
ax.set_xlabel("search time, seconds (log scale)")
ax.set_ylabel("llm_score")
ax.set_title("Molecule-analysis: search time vs. score")
plt.show()
```

```python
mol_cat_stats = (
    mol_pool.groupby("llm_category", dropna=False)
            .agg(n=("SMILES", "size"), mean_score=("llm_score", "mean"),
                 mean_confidence=("llm_confidence", "mean"), mean_agreement=("llm_category_agreement", "mean"))
            .reset_index()
)
mol_cat_stats["pct"] = mol_cat_stats["n"] / mol_cat_stats["n"].sum() * 100
mol_cat_stats = (mol_cat_stats.assign(_o=mol_cat_stats["llm_category"].map(CAT_RANK))
                               .sort_values("_o").drop(columns="_o").round(1).reset_index(drop=True))
mol_cat_stats
```

```python
for cat in MOL_CATS:
    cat_df = mol_pool[mol_pool["llm_category"] == cat]
    if len(cat_df) == 0:
        continue
    analyze_category(cat_df, "molecule_only", cat)
```
