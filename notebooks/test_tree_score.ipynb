```python
import ast
from dataclasses import dataclass
from typing import Any, Dict, List, Optional, Sequence, Set, Tuple

@dataclass
class RouteTree:
    """A parsed retrosynthesis route as a molecule directed acyclic graph."""

    target: Optional[str]
    reactions: List[Tuple[str, List[str], str, int]]  # (product, reactants, label, depth)
    products: Set[str]
    reactants: Set[str]

    @property
    def leaves(self) -> Set[str]:
        """Building blocks: molecules that appear as a reactant but never a product."""
        return self.reactants - self.products

    @property
    def intermediates(self) -> Set[str]:
        """Non-target products (the internal nodes of the tree)."""
        return self.products - ({self.target} if self.target else set())


def parse_route(route: Dict[Any, Sequence[Sequence[str]]]) -> RouteTree:
    """Turn a ``{depth: [['P => R1.R2', label], ...]}`` dict into a :class:`RouteTree`.

    Reactants are split on ``.``, which also separates salt/mixture components in
    SMILES; this is acceptable for structural counting but may slightly over-count
    the rare salt.
    """
    reactions: List[Tuple[str, List[str], str, int]] = []
    products: Set[str] = set()
    reactants: Set[str] = set()
    for depth, steps in route.items():
        for entry in steps:
            reaction: str = entry[0]
            label: str = entry[1] if len(entry) > 1 else ""
            if "=>" not in reaction:
                continue
            product_part, reactant_part = reaction.split("=>", 1)
            product: str = product_part.strip()
            these_reactants: List[str] = [
                token.strip() for token in reactant_part.strip().split(".") if token.strip()
            ]
            reactions.append((product, these_reactants, label, int(depth)))
            products.add(product)
            reactants.update(these_reactants)

    roots: Set[str] = products - reactants
    target: Optional[str] = next(iter(roots)) if roots else None
    return RouteTree(target, reactions, products, reactants)

route = df.iloc[0]["route"]
route = ast.literal_eval(route) if route.strip() else {}
tree = parse_route(route)
print(route)
print(tree.reactants)
print(tree.products)
print()
for r in tree.reactions:
    print(r)
```

```python
import networkx as nx

def route_to_nx(route: str | Dict[Any, Sequence[Sequence[str]]]) -> nx.DiGraph:
    """Build a networkx DiGraph directly from a raw route dict.

    ``route`` is a ``{depth: [['P => R1.R2', label], ...]}`` mapping. Molecules are
    nodes; edges point from a product to each of its reactants (the retrosynthesis
    expansion direction, so the target is the root and building blocks are leaves)
    and carry the reaction ``label`` and ``depth``. Reactants are split on ``.``,
    which also separates salt/mixture components in SMILES; this is acceptable for
    structural counting but may slightly over-count the rare salt. Each node carries
    ``depth`` (the depth at which it's produced, ``None`` for the target) and
    ``kind`` in {"target", "intermediate", "leaf"}.
    """
    if isinstance(route, str):
        route = ast.literal_eval(route) if route.strip() else {}
    g = nx.DiGraph()
    depths: Dict[str, int] = {}
    for depth, steps in route.items():
        for entry in steps:
            reaction: str = entry[0]
            label: str = entry[1] if len(entry) > 1 else ""
            if "=>" not in reaction:
                continue
            product_part, reactant_part = reaction.split("=>", 1)
            product: str = product_part.strip()
            reactants: List[str] = [
                token.strip() for token in reactant_part.strip().split(".") if token.strip()
            ]
            g.add_node(product)
            depths[product] = min(int(depth), depths.get(product, int(depth)))
            for reactant in reactants:
                g.add_edge(product, reactant, label=label, depth=int(depth))

    for molecule in g.nodes:
        if g.in_degree(molecule) == 0:
            kind = "target"
        elif g.out_degree(molecule) == 0:
            kind = "leaf"
        else:
            kind = "intermediate"
        g.nodes[molecule]["kind"] = kind
        g.nodes[molecule]["depth"] = depths.get(molecule)

    return g
```

```python
def hierarchy_pos(g: nx.DiGraph, root: str, width: float = 1.0, vert_gap: float = 0.2) -> Dict[str, Tuple[float, float]]:
    """Top-down tree layout, root at the top, children spread out below.

    ``nx.draw``'s default ``spring_layout`` is force-directed and ignores
    parent/child structure, so even a genuine tree looks tangled. Assumes ``g`` is
    an out-arborescence reachable from ``root`` via directed edges (as produced by
    :func:`route_to_nx`). Adapted from the standard networkx "hierarchy_pos" recipe
    (https://stackoverflow.com/a/29597209) since networkx has no built-in tree layout.
    """
    def _walk(node: str, left: float, right: float, depth: int, pos: Dict[str, Tuple[float, float]]) -> None:
        pos[node] = ((left + right) / 2, -depth * vert_gap)
        children = list(g.successors(node))
        if not children:
            return
        step = (right - left) / len(children)
        for i, child in enumerate(children):
            _walk(child, left + i * step, left + (i + 1) * step, depth + 1, pos)

    pos: Dict[str, Tuple[float, float]] = {}
    _walk(root, 0, width, 0, pos)
    return pos


pos = hierarchy_pos(g, target)
nx.draw(g, pos=pos, with_labels=False, arrows=True, node_size=200)
```

```python
def _longest_linear_sequence(tree: RouteTree) -> int:
    """Return the longest root-to-leaf path counted in reactions."""
    children: Dict[str, List[str]] = {}
    for product, reactants, _label, _depth in tree.reactions:
        children.setdefault(product, []).extend(reactants)

    memo: Dict[str, int] = {}

    def depth_of(node: str, seen: frozenset) -> int:
        if node not in children or node in seen:
            return 0
        if node in memo:
            return memo[node]
        result = 1 + max(depth_of(child, seen | {node}) for child in children[node])
        memo[node] = result
        return result

    return depth_of(tree.target, frozenset()) if tree.target else 0

print("longest linear sequence:", _longest_linear_sequence(tree))
nx.dag_longest_path_length(g)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import Counter

# -------------------------------------------------------------
# 1. Basic Structural Metrics (Unrooted / General Trees)
# -------------------------------------------------------------

def leaf_to_node_ratio(T: nx.Graph) -> float:
    """Computes the ratio of leaves (degree <= 1) to total nodes."""
    n = T.number_of_nodes()
    if n <= 1:
        return 1.0
    leaves = [node for node, deg in T.degree() if deg == 1]
    return len(leaves) / n


def diameter_to_size_ratio(T: nx.Graph) -> float:
    """Computes the ratio of the tree's diameter to total nodes."""
    G = T.to_undirected() if T.is_directed() else T
    n = G.number_of_nodes()
    if n <= 1:
        return 0.0
    diameter = nx.diameter(G)
    return diameter / n


def wiener_index(T: nx.Graph) -> int:
    G = T.to_undirected() if T.is_directed() else T
    all_lengths = dict(nx.all_pairs_shortest_path_length(G))
    return sum(
        dist
        for u, lengths in all_lengths.items()
        for v, dist in lengths.items()
        if u < v
    )

def normalized_wiener_star(T: nx.Graph) -> float:
    """
    Normalizes by the theoretical minimum (Star Graph).
    Returns a score >= 1.0, where 1.0 is maximally branchy/compact.
    """
    G = T.to_undirected() if T.is_directed() else T
    n = G.number_of_nodes()
    
    if n <= 2:
        return 1.0
        
    w = wiener_index(G)
    w_min = (n - 1) ** 2
    return w / w_min

def normalized_wiener_minmax(T: nx.Graph) -> float:
    """
    Scales the Wiener index strictly between 0.0 and 1.0:
    - 0.0: Star Graph (maximally compact/branchy)
    - 1.0: Linear Path (minimally branchy/elongated)
    """
    G = T.to_undirected() if T.is_directed() else T
    n = G.number_of_nodes()
    
    if n <= 2:
        return 0.0
        
    w = wiener_index(G)
    w_min = (n - 1) ** 2
    w_max = (n * (n**2 - 1)) / 6
    
    if w_max == w_min:
        return 0.0
        
    return (w - w_min) / (w_max - w_min)

# -------------------------------------------------------------
# 2. Phylogenetic / Balance Indices (Rooted Directed Trees)
# -------------------------------------------------------------

def colless_index(T_directed: nx.DiGraph) -> int:
    """
    Calculates the Colless Index for a strictly binary rooted tree.
    T_directed must be a DAG / out-tree with edges pointing from parent to child.
    """
    # Find root (node with in-degree 0)
    roots = [n for n, d in T_directed.in_degree() if d == 0]
    if not roots:
        raise ValueError("Graph must have a clear root (in-degree 0).")
    
    # Precompute subtree sizes (number of leaves in subtree)
    leaf_counts = {}
    for node in reversed(list(nx.topological_sort(T_directed))):
        children = list(T_directed.successors(node))
        if not children:
            leaf_counts[node] = 1
        else:
            leaf_counts[node] = sum(leaf_counts[c] for c in children)

    colless = 0
    for node in T_directed.nodes():
        children = list(T_directed.successors(node))
        if len(children) == 2:
            left_size = leaf_counts[children[0]]
            right_size = leaf_counts[children[1]]
            colless += abs(left_size - right_size)
            
    return colless


def sackin_index(T_directed: nx.DiGraph) -> int:
    """
    Computes the Sackin Index: sum of root-to-leaf depths for all leaves.
    """
    roots = [n for n, d in T_directed.in_degree() if d == 0]
    if not roots:
        raise ValueError("Graph must have a clear root (in-degree 0).")
    root = roots[0]
    
    depths = nx.single_source_shortest_path_length(T_directed, root)
    leaves = [n for n, d in T_directed.out_degree() if d == 0]
    
    return sum(depths[leaf] for leaf in leaves)


# -------------------------------------------------------------
# 3. Hierarchical Flow Metrics (Horton–Strahler & Bifurcation)
# -------------------------------------------------------------

def horton_strahler_orders(T_directed: nx.DiGraph) -> dict:
    """
    Computes Horton-Strahler stream orders for every node.
    Leaves start at order 1.
    """
    orders = {}
    # Process from leaves to root using reverse topological sort
    for node in reversed(list(nx.topological_sort(T_directed))):
        children = list(T_directed.successors(node))
        if not children:
            orders[node] = 1
        else:
            child_orders = [orders[c] for c in children]
            max_order = max(child_orders)
            # If two or more children share the maximum order, increment by 1
            if child_orders.count(max_order) >= 2:
                orders[node] = max_order + 1
            else:
                orders[node] = max_order
                
    return orders


def bifurcation_ratios(T_directed: nx.DiGraph) -> dict:
    """
    Computes the Horton bifurcation ratio (R_B = N_i / N_{i+1}) between consecutive orders.
    """
    orders = horton_strahler_orders(T_directed)
    counts = Counter(orders.values())
    max_order = max(counts.keys())
    
    ratios = {}
    for i in range(1, max_order):
        if counts[i + 1] > 0:
            ratios[f"R_{i}/{i+1}"] = counts[i] / counts[i + 1]
            
    return ratios


for i in range(10):
    if not df.iloc[i]["resolved"]:
        continue

    print(f"Route {i+1}")
    route = df.iloc[i]["route"]
    route = ast.literal_eval(route) if route.strip() else {}
    g = route_to_nx(route)

    target = next(n for n, d in g.in_degree() if d == 0)
    pos = hierarchy_pos(g, target)
    plt.figure(figsize=(8, 6))
    nx.draw(g, pos=pos, with_labels=False, arrows=True, node_size=100)
    plt.show()
    plt.close()

    print("Leaf-to-Node Ratio:", leaf_to_node_ratio(g))
    print("Diameter / Size:   ", diameter_to_size_ratio(g))
    print("Wiener Index:      ", wiener_index(g))
    print("Normalized Wiener: ", normalized_wiener_star(g))
    print("Normalized Wiener (MinMax):", normalized_wiener_minmax(g))
    print("Colless Index:     ", colless_index(g))
    print("Sackin Index:      ", sackin_index(g))
    # print("Strahler Orders:   ", horton_strahler_orders(g))
    print("Bifurcation Ratios:", bifurcation_ratios(g))
    print('-' * 80)
```

```python
def route_graph_metrics(route: str | Dict[Any, Sequence[Sequence[str]]]) -> Dict[str, float]:
    """Compute the graph-based route metrics defined above for a single route.

    Returns NaN-filled defaults for routes with no reactions (e.g. unresolved or
    single-molecule routes), and collapses ``bifurcation_ratios``' variable-length
    dict into two scalar summaries (mean ratio, number of Strahler orders) so the
    result is a fixed-width row usable as new dataframe columns.
    """
    g = route_to_nx(route)

    metrics: Dict[str, float] = {
        "graph_n_nodes": g.number_of_nodes(),
        "graph_leaf_to_node_ratio": float("nan"),
        "graph_diameter_to_size": float("nan"),
        "graph_wiener_index": float("nan"),
        "graph_wiener_norm_star": float("nan"),
        "graph_wiener_norm_minmax": float("nan"),
        "graph_colless_index": float("nan"),
        "graph_sackin_index": float("nan"),
        "graph_n_strahler_orders": float("nan"),
        "graph_mean_bifurcation_ratio": float("nan"),
    }

    if g.number_of_nodes() == 0:
        return metrics

    metrics["graph_leaf_to_node_ratio"] = leaf_to_node_ratio(g)
    metrics["graph_diameter_to_size"] = diameter_to_size_ratio(g)
    metrics["graph_wiener_index"] = wiener_index(g)
    metrics["graph_wiener_norm_star"] = normalized_wiener_star(g)
    metrics["graph_wiener_norm_minmax"] = normalized_wiener_minmax(g)
    metrics["graph_colless_index"] = colless_index(g)
    metrics["graph_sackin_index"] = sackin_index(g)

    ratios = bifurcation_ratios(g)
    orders = horton_strahler_orders(g)
    metrics["graph_n_strahler_orders"] = max(orders.values()) if orders else float("nan")
    metrics["graph_mean_bifurcation_ratio"] = (
        sum(ratios.values()) / len(ratios) if ratios else float("nan")
    )

    return metrics


graph_metrics_df = df["route"].apply(route_graph_metrics).apply(pd.Series)
df = pd.concat([df, graph_metrics_df], axis=1)
df[graph_metrics_df.columns].describe()
```
